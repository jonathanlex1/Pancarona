## Data Cleaning 

- terdapat inconsistent tipe data pada totalcharges yang seharus adalah bertipe float dan SeniorCitizen yang seharus bertipe object
- inconsistent value (No internet service yang dapat disatukan saja menjadi No dan juga no phone service dapat diganti dengan no saja)
- inconesistent huruf kapital pada kolom kolom
- menambahakan kolom baru yang memungkinkan

### Columns

```python
#rename columns 

df.columns = [col.lower() for col in df.columns]

col_rename = {
    'customerid' : 'customer_id',
    'seniorcitizen' : 'senior_citizen',
    'phoneservice' : 'phone_service',	
    'multiplelines': 'multiple_lines',	
    'internetservice' : 'internet_service',
    'onlinesecurity' : 'online_security',
    'onlinebackup' : 'online_backup',
    'deviceprotection' : 'device_protection',
    'techsupport' : 'tech_support', 
    'streamingtv' : 'streaming_tv',
    'streamingmovies' : 'streaming_movies',
    'paperlessbilling' : 'paperless_billing',	
    'paymentmethod' : 'payment_method',	
    'monthlycharges' : 'monthly_charges',
    'totalcharges' : 'total_charges'	
}
df = df.rename(columns=col_rename)

df.head()
```

### Values of data

```python
def values_check(df) :  
    for col in df.select_dtypes(include='object').columns : 
        print(col) 
        print(df[col].unique())
```

```python
df['senior_citizen'].unique()
```

```python
#replacing the values of some columns
df['senior_citizen'] = df['senior_citizen'].replace(0, "No").replace(1, 'Yes')
df['multiple_lines'] = df['multiple_lines'].replace('No phone service', 'No')
df['online_security'] = df['online_security'].replace('No internet service', 'No')
df['online_backup'] = df['online_backup'].replace('No internet service', 'No')
df['device_protection'] = df['device_protection'].replace('No internet service', 'No')
df['tech_support']= df['tech_support'].replace('No internet service', 'No')
df['streaming_tv'] = df['streaming_tv'].replace('No internet service', 'No')
df['streaming_movies'] = df['streaming_movies'].replace('No internet service', 'No')
```

```python
#changing type data of totalcharges column
df['total_charges'] = pd.to_numeric(df['total_charges'], errors='coerce')
```

```python
#cleaning null data 
df['total_charges'] = df['total_charges'].fillna(0)
```

### Add new columns 

```python
#add some columns 
df['expected_charges'] = (df['monthly_charges'] * df['tenure']).round(2)
df['diff_charges'] = (df['total_charges'] - df['expected_charges']).round(2)
```

```python
df['total_services'] = df[['phone_service','multiple_lines','internet_service','online_security','online_backup', 'device_protection', 'tech_support', 'streaming_tv', 'streaming_movies']].replace({'Yes' : 1, 'No' : 0, 'DSL':1, 'Fiber optic':1}).sum(axis=1)
```
